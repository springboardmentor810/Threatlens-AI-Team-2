```python
import pandas as pd
```

```python
summary0 = pd.read_csv("../../reports/summaries/train_0_summary.csv")
summary1 = pd.read_csv("../../reports/summaries/train_1_summary.csv")
summary2 = pd.read_csv("../../reports/summaries/train_2_summary.csv")
summary3 = pd.read_csv("../../reports/summaries/train_3_summary.csv")
summary4 = pd.read_csv("../../reports/summaries/train_4_summary.csv")
summary5 = pd.read_csv("../../reports/summaries/train_5_summary.csv")
```

```python
final_summary = pd.concat(
    [summary0, summary1, summary2,summary3, summary4, summary5],
    ignore_index=True
)

final_summary
```

```python
import matplotlib.pyplot as plt

final_summary.plot(
    x="Dataset",
    y="Rows",
    kind="bar",
    figsize=(8,5)
)

plt.title("Rows in Each EMBER Dataset")
plt.ylabel("Number of Rows")
plt.xticks(rotation=0)
plt.show()
```

```python
import matplotlib.pyplot as plt

final_summary.plot(
    x="Dataset",
    y="Memory (MB)",
    kind="bar",
    figsize=(8,5)
)

plt.title("Memory Usage in Each EMBER Dataset")
plt.ylabel("Memory (MB)")
plt.xticks(rotation=0)
plt.show()
```

```python
import matplotlib.pyplot as plt

final_summary.plot(
    x="Dataset",
    y="Malware Samples",
    kind="bar",
    figsize=(8,5)
)

plt.title("Malware Samples in Each EMBER Dataset")
plt.ylabel("Number of Malware Samples")
plt.xticks(rotation=0)
plt.show()
```
